```python
nonull_hotel_reservation = hotel_reservation.drop(hotel_reservation[(hotel_reservation["no_of_weekend_nights"] == 0) & (hotel_reservation["no_of_week_nights"] == 0)].index)
nonull_hotel_reservation
```

```python
nonull_hotel_reservation.isnull().sum()
```

```python
#열 정제
new_hotel_reservation = nonull_hotel_reservation.drop(["Booking_ID", "room_type_reserved", "no_of_special_requests", "lead_time", "arrival_date", "arrival_year"], axis=1)
new_hotel_reservation['total_nights'] = new_hotel_reservation['no_of_weekend_nights'] + new_hotel_reservation['no_of_week_nights']
new_hotel_reservation
```

```python
new_hotel_reservation['type_of_meal_plan'].unique()
```

```python
new_hotel_reservation['type_of_meal_plan'] = new_hotel_reservation['type_of_meal_plan'].replace({'Not Selected': 0, 'Meal Plan 1': 1, 'Meal Plan 2': 1, 'Meal Plan 3': 1})
```

```python
new_hotel_reservation['booking_status'].unique()
```

```python
new_hotel_reservation['booking_status'] = np.where(new_hotel_reservation['booking_status'] == 'Canceled', 0, 1)
```

```python
new_hotel_reservation['market_segment_type'].unique()
```

```python
new_hotel_reservation['market_segment_type'] = new_hotel_reservation['market_segment_type'].replace({'Offline': 0, 'Online': 1, 'Corporate': 2,  'Aviation': 3, 'Complementary': 4})
```

# 시각화




```python
sns.displot(final_hotel_reservation,x="avg_price_per_room",hue="booking_status", kind="kde")
plt.xlabel("Average Price per Room")
plt.ylabel("Booking Status")
plt.title("Average Price per Room")
```

```python
not_canceled_reservation = final_hotel_reservation[final_hotel_reservation['booking_status'] == 1]
average_price_by_month = not_canceled_reservation.groupby('arrival_month')['avg_price_per_room'].mean()

plt.figure(figsize=(11, 7))
ax = average_price_by_month.plot(kind='line')
plt.xlabel('Month')
plt.ylabel('Average Price')
plt.title('Average Price by Month')
for i, j in enumerate(average_price_by_month):
    ax.text(i, j, round(j, 2), ha='center', va='bottom', fontsize=9)
    plt.xticks(rotation = 0)
plt.show()
```

```python
final_hotel_reservation['no_of_total'] = final_hotel_reservation['no_of_adults'] + final_hotel_reservation['no_of_children']
plt.figure(figsize=(10, 6))
n, bins, patches = plt.hist(final_hotel_reservation['no_of_total'], bins=range(1, 15), edgecolor='white', color = 'red', alpha=0.7)
plt.xlabel('Total Guests')
plt.ylabel('Count')
plt.title('Distribution of Number of Guests')
plt.xticks(range(1, 15))
for i in range(len(patches)):
    plt.text(patches[i].get_x() + patches[i].get_width() / 2, patches[i].get_height(), int(n[i]), ha='center', va='bottom')
plt.show()
```

# 장기 투숙 분석

```python
correlation = final_hotel_reservation['total_nights'].corr(final_hotel_reservation['avg_price_per_room'])
print(f'Average Price & Total Nights Correlation: {correlation:.2f}')
```

```python
correlation, p_value = pearsonr(final_hotel_reservation['total_nights'], final_hotel_reservation['avg_price_per_room'])
print(f'Average Price & Total Nights P-value: {p_value:.4f}')
```

```python
final_hotel_reservation['price_category'] = pd.qcut(final_hotel_reservation['avg_price_per_room'], q = 5, labels = [
    'Very Cheap', 'Cheap', 'Moderate', 'Expensive', 'Very Expensive'])
average_nights_by_category = final_hotel_reservation.groupby('price_category')['total_nights'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='price_category', y='total_nights', data=average_nights_by_category)
plt.xlabel('Price Category')
plt.ylabel('Average Total Nights')
plt.title('Average Total Nights by Price Category')
plt.ylim(2, 3.3, 0.2)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(final_hotel_reservation['total_nights'], bins=30, kde=True, color='skyblue')
plt.xlabel('Total Nights')
plt.ylabel('Frequency')
plt.title('Distribution of Total Nights')
plt.show()
```

```python
final_hotel_reservation['stay_category'] = pd.cut(final_hotel_reservation['total_nights'], bins=[-float('inf'), 6, float('inf')], labels=['Short Stay', 'Long Stay'])
stay_category_counts = final_hotel_reservation['stay_category'].value_counts()
long_stay_ratio = stay_category_counts.get('Long Stay', 0) / len(final_hotel_reservation)
short_stay_ratio = stay_category_counts.get('Short Stay', 0) / len(final_hotel_reservation)
labels = ['Long Stay', 'Short Stay']
sizes = [long_stay_ratio, short_stay_ratio]
colors = ['lightcoral', 'lightskyblue']

plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=140)
plt.title('Distribution of Stay Categories')
plt.show()
```

```python
long_stay_data = final_hotel_reservation[final_hotel_reservation['stay_category'] == 'Long Stay']
correlation, p_value = pearsonr(long_stay_data['avg_price_per_room'], long_stay_data['total_nights'])

print(f'Correlation: {correlation:.2f}')
print(f'P-value: {p_value:.4f}')
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(long_stay_data['avg_price_per_room'], bins=30, kde=True, color='blue')
plt.xlabel('Average Price per Room')
plt.ylabel('Frequency')
plt.title('Distribution of Average Price per Room for Long Stay')
plt.show()
```

```python
long_stay_by_price_category = long_stay_data.groupby('price_category').size().reset_index(name='long_stay')

plt.figure(figsize=(10, 6))
sns.barplot(x='price_category', y='long_stay', data = long_stay_by_price_category, palette='viridis')
plt.xlabel('Price Category')
plt.ylabel('Number of Long Stays')
plt.title('Number of Long Stays by Price Category')
plt.show()
```
